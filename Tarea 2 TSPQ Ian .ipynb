#2.1 **Bucle** for: Empleado en procesamiento de listas


1. Crea una funcion que permita eliminar los valores nulos (None) de los valores numericos de las siguientes listas.

```python
# Paso 1, colocar las listas, de las que se busca eliminar todos aquellos elementos "None"
Lista_1 = [7.2, 7.8, 6.8, 8.0, None, '8.2', 5.6, 8.2, 7.7, 7.5, None, 5.8]
Lista_2 = ['6.8', None, 6.8, 6.1, 7.9, 9.4, None]
Lista_3 = [8.5, '9.0', None, 7.7]

# Paso 2, quitar los valores None de cada lista

Lista_1_limpia = []     # Se debera de crear una nueva lista, la cual contendra los elementos una vez se borren los "None", a esta se le denomiara limpia
for i in Lista_1:       # Se abre el ciclo for, para realizar la accion para la lista 1
  if i is not None:     # Se coloca un condicional para que todas aquellas variables que no sean None sean consideradas
    Lista_1_limpia.append(i)  # Se agrega append para que se agregue todo aquel que no sea None en el siguinte valor de la lista


Lista_2_limpia = []
for i in Lista_2:
  if i is not None:
    Lista_2_limpia.append(i)

Lista_3_limpia = []
for i in Lista_3:
  if i is not None:
    Lista_3_limpia.append(i)

print ('Lista 1 limpia: ',  Lista_1_limpia)     # Se pide que se imprima cada lista, junto a su texto correspondiente
print ('Lista 2 limpia: ', Lista_2_limpia)
print ('Lista 3 limpia: ', Lista_3_limpia)
```

2. Utilizando un bucle, ordena todos los valores numéricos obtenidos en el inciso anterior de menor a mayor y divídelos en tres listas del mismo tamaño.

```python
# Paso 1, juntar las listas

Lista_4 = Lista_1_limpia + Lista_2_limpia + Lista_3_limpia

# Paso 2, convertir todos los datos a numeros

for i in range(len(Lista_4)):
  Lista_4[i] = float(Lista_4[i])


#Paso 3, ordenarlas de menor a mayor

Lista_4.sort()

# Paso 4, dividirlos en 3 lineas del mismo tamaño

Lista_4_1 = Lista_4[0:6]
Lista_4_2 = Lista_4[6:12]
Lista_4_3 = Lista_4[12:18]

print ('Lista 1 nueva: ', Lista_4_1)
print ('Lista 2 nueva: ', Lista_4_2)
print ('Lista 3 nueva: ', Lista_4_3)
```

3. Crea un programa que reciba del usuario los nueve elementos de una matriz
cuadrada de tamaño 3 x 3 y calcule el determinante de dicha matriz.

a)Usando la regla de Sarrus

b)  Resuelve utilizando la paquetería NumPy para el cálculo del
determinante


```python
# Paso 1, colocar un elemento vacio, mismo que sera aquel que se llene de datos, junto a la instruccion para el llenado

matriz = []
print("Ingrese los 9 elementos de la matriz 3x3 (fila por fila): ")

# Paso 2, Abrir el ciclo for para la creacion de la matriz

for i in range(3):    # Donde se deberan colocar 2 ciclos, dado que uno sera para las filas y otro para las columnas
  fila = []
  for j in range(3):
    while True:

# Paso 3, colocar un input que funciona para añadir los elementos de la matriz

      try:
        valor = float(input(f"Ingrese el elemento [{i+1}][{j+1}]: "))     #Este se colocara como tipo float, para que detecte cualquier tipo de numero
        fila.append(valor)      # Con append se agregan como ultimo dato
        break     # Para que detecte el fin de cada elemento de la matriz

# Paso 4, colocar una exepcion de datos, para segurarse que solo se coloquen numeros.
      except ValueError:
        print("Entrada incorrecta. Por favor, ingrese un número")
  matriz.append(fila)

print("\nSu matriz es: ")
for fila in matriz:
    print(fila)

# Inciso b), Calcular el determinante de la matriz usando la regla de Sarrus

a, b, c = matriz[0][0], matriz[0][1], matriz[0][2]    # Se establecen variables para cada valor de la matriz colocada anteriormente
d, e, f = matriz[1][0], matriz[1][1], matriz[1][2]
g, h, i = matriz[2][0], matriz[2][1], matriz[2][2]



determinante = (a*e*i + b*f*g + c*d*h) - (c*e*g + b*d*i + a*f*h)    # Se calcula el determinante usando la regla de Sarrus

print(f"\nEl determinante de la matriz usando la regla de Sarrus es: {determinante}")
```

4. Pide al usuario dos listas de números reales de la misma longitud y calcula su
producto escalar utilizando un bucle for.

Nota: Recuerda que el producto escalar se
define como la suma de los productos de los elementos correspondientes.

```python
# Función para obtener una lista de números reales que sean colocados por el usuario

def Listareal(a):     # Se define una lista a
    while True:       # Para que el ciclo sea infinito
        try:
            inputreal = input(a)    #Se coloca el input para que se empiece a llenar la lista a

            # Convertir la cadena de entrada en una lista de números reales
            Lista = [float(x.strip()) for x in inputreal.split(',')]    # Se define la lista, misma que se encontrara separada por comas
            return Lista
        except ValueError:
            print("Entrada inválida. Por favor, ingrese números reales separados por comas.")

# Pedir al usuario las dos listas
lista_1 = []
lista_2 = []

while True:
    lista_1 = Listareal("Ingrese los números de la primera lista, separados por comas (ej. 1.0, 2.5, 3.0): ")
    lista_2 = Listareal("Ingrese los números de la segunda lista, separados por comas (ej. 4.0, 5.5, 6.0): ")

    if len(lista_1) == len(lista_2):    # Para asegurarse que ambas listas tengan la misma cantidad de valores
        break
    else:
        print("Las listas deben tener la misma longitud. Inténtelo de nuevo.")

# Calcular el producto escalar usando un bucle for

producto_escalar = 0
for i in range(len(lista_1)):
    producto_escalar += lista_1[i] * lista_2[i]   # No olvidar que el + antes del =, es para que realice una sumatoria

print(f"\nLa primera lista ingresada es: {lista_1}")
print(f"La segunda lista ingresada es: {lista_2}")
print(f"El producto escalar de las dos listas es: {producto_escalar}")
```

#2.3 **Bucle** for:  Empleado en cadenas de texto

5. Crea un programa que reciba una cadena de caracteres y determine si el texto
ingresado corresponde al nombre de un estado o a una capital de la República
Mexicana

```python
import unicodedata      # Para trabajar de mejor manera con texto especial, es decir, acentos, mayusculas, etc.

def Texto_simple(T):    # Definiendo una funcion para convertir cualquier texto en uno mas simple.

    # Para eliminar acentos y evitar problemas al colocar texto:

    T = unicodedata.normalize('NFKD', T).encode('ascii', 'ignore')  # La variable T, se redefine.
# Donde 'NFKD' descomponera los acentos en letras normales y encode, hara que se pueda tratar todo como tipo str.

    return T.lower()      # Lower, hara que no importa si se coloca todo en mayusculas o misculas, podra detectar el texto de cualquier forma.

# Listas de estados y capitales de México
estados = [
    "Aguascalientes", "Baja California", "Baja California Sur", "Campeche", "Chiapas",
    "Chihuahua", "Coahuila", "Colima", "Ciudad de México", "Durango",
    "Estado de México", "Guanajuato", "Guerrero", "Hidalgo", "Jalisco", "Michoacán",
    "Morelos", "Nayarit", "Nuevo León", "Oaxaca", "Puebla", "Querétaro", "Quintana Roo",
    "San Luis Potosí", "Sinaloa", "Sonora", "Tabasco", "Tamaulipas", "Tlaxcala",
    "Veracruz ", "Yucatán", "Zacatecas"
]

capitales = [
    "Aguascalientes", "Mexicali", "La Paz", "San Francisco de Campeche", "Tuxtla Gutiérrez",
    "Chihuahua", "Saltillo", "Colima", "Ciudad de México", "Durango",
    "Toluca de Lerdo", "Guanajuato", "Chilpancingo", "Pachuca", "Guadalajara", "Morelia",
    "Cuernavaca", "Tepic", "Monterrey", "Oaxaca", "Puebla", "Santiago de Querétaro", "Chetumal",
    "San Luis Potosí", "Culiacán ", "Hermosillo", "Villahermosa", "Ciudad Victoria", "Tlaxcala de Xicohténcatl",
    "Xalapa-Enríquez", "Mérida", "Zacatecas"
    ]

# Normalizar listas para una comparación más fácil

estados_normalizados = [Texto_simple(e) for e in estados]
capitales_normalizados = [Texto_simple(c) for c in capitales]

# Pedir al usuario una cadena de caracteres
texto_usuario = input("Ingrese un nombre de un estado o capital de México: ")
texto_usuario_normalizado = Texto_simple(texto_usuario)


# Determinar si es un estado o capital
Destado = texto_usuario_normalizado in estados_normalizados
Dcapital = texto_usuario_normalizado in capitales_normalizados

# Colocar un condicional para que se pueda saber si se coloco, o no un estado, una capital o ambios.

if Destado and Dcapital:
    print(f"'{texto_usuario}' es un estado y también la capital de un estado en la República Mexicana.")
elif Destado:
    print(f"'{texto_usuario}' corresponde a un estado de la República Mexicana.")
elif Dcapital:
    print(f"'{texto_usuario}' corresponde a una capital de la República Mexicana.")
else:
    print(f"'{texto_usuario}' no corresponde a un estado ni a una capital de la República Mexicana, volver a intentar.")
```

#2.4 **Bucle** for:  Empleado en control de medición

6. Utiliza un bucle while para solicitar números al usuario y agregarlos a una lista.

El programa deberá detenerse cuando la suma total de los elementos sea mayor
que 100. Al finalizar, muestra la lista completa y la suma total.

```python
# Paso 1. Establecer una lista vacia
numeros = []

total = 0   # Colocar un valor inicial


print("Ingrese una serie de numeros hasta alcanzar 100")

# Colocar un condicional para limitar el maximo.

while total <= 100:
    try:
        N = float(input("Ingrese un número: "))   #Colocar el input, donde puedn ser cualquier numero
        numeros.append(N)
        total += N
        print(f"Suma actual: {total}")

    except ValueError:        # Colocar una limitante
        print("Entrada inválida. Por favor, ingrese un número válido.")

print(f"\nLista completa: {numeros}")
print(f"Suma total: {total}")
```

#2.5 **Bucle** for: Empleado en menu de opciones

7. Crea una lista vacía y desarrolla un programa que muestre un menú de opciones al
usuario utilizando un bucle while.

El menú deberá permitir realizar las siguientes acciones:

-Agregar un número a la lista.

-Eliminar el último número de la lista.

-Mostrar el contenido actual de la lista.

-Salir del programa.

El programa deberá repetirse mientras el usuario no seleccione la opción de
salir. Para ello, utiliza una variable de control que permita terminar el bucle while
cuando el usuario así lo indique.

```python
# Paso 1. Crear lista vacia
lista = []
opcion = ""     # Se coloca para almacenar la seleccion del usuario en el menu

#Paso 2. Inicial el bucle

while opcion != "4":      #Limitar el bucle hasta alcanzar la opcion 4
    print("\nMenú:")
    print("1. Agregar número")
    print("2. Eliminar último número")
    print("3. Mostrar lista")
    print("4. Salir")

    opcion = input("Seleccione una opción: ")

# Paso 3. Crear la opcion 1, para agregar numeros

    if opcion == "1":
        try:
            numero = float(input("Ingrese el número a agregar: "))
            lista.append(numero)
            print(f"'{numero}' agregado a la lista.")
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número válido.")

# Paso 4. Crear la opcion 2, para eliminar el ultimo numero

    elif opcion == "2":
        if lista:
            numero_eliminado = lista.pop()
            print(f"'{numero_eliminado}' eliminado de la lista.")
        else:
            print("La lista está vacía, no hay elementos para eliminar.")

# Paso 5. Crear la opcion 3, para mostrar la lista de numeros

    elif opcion == "3":
        if lista:
            print(f"Contenido actual de la lista: {lista}")
        else:
            print("La lista está vacía.")

  # Paso 6. Crear la opcion 4, Para salir

    elif opcion == "4":
        print("Saliendo del programa. ¡Hasta luego!")
    else:
        print("Opción no válida. Por favor, intente de nuevo.")
```
